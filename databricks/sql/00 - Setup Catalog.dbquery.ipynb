```sql
-- A catalog is th top level of Unity Catalog's 3-level namespace: catalog.schema.table
CREATE CATALOG IF NOT EXISTS lumora
    COMMENT 'Lumora Skincare - marketing & ecommerce analytics (portifolio project)';

USE CATALOG lumora;

-- One schema per medallion layer: permissions and lienage are managed per schema
CREATE SCHEMA IF NOT EXISTS raw COMMENT 'landing files (volumes), no tables';
CREATE SCHEMA IF NOT EXISTS bronze COMMENT 'Raw-faithful Delta tables, 1:1 with source files';
CREATE SCHEMA IF NOT EXISTS silver COMMENT 'Cleaned, typed, deduplicated, conformed';
CREATE SCHEMA IF NOT EXISTS gold COMMENT 'Star Schema  + business marts';

-- A MNAGED VOLUME = governed file storage inside UC (the modern replacement for DBFS).
-- FILES here get a path like /Volumes/lumora/raw/landing/...

CREATE VOLUME IF NOT EXISTS raw.landing
 COMMENT 'Raw source exports: ads, web sessions, order, CRM';

CREATE VOLUME IF NOT EXISTS lumora.raw.checkpoints
  COMMENT 'Auto Loader checkpoints and schema locations';

SHOW VOLUMES IN raw;
```
